### Decision: possibly auto-converted titles
Found 5 titles matching a date-like pattern: 9-Feb, 15-Aug, 22-Jul, SPF-18, 1-Oct.
SPF-18 looks like a genuine title (false positive). The other 4 were probably
converted from names such as "22 July" by spreadsheet software. The original
names cannot be recovered from the file alone, so titles were left unchanged and
the issue is documented as a limitation. Impact: 4 of 8,787 rows (about 0.05%).

```python
print(df_clean.shape)
print(df_clean["title"].isin(["9-Feb", "15-Aug", "22-Jul", "SPF-18", "1-Oct"]).sum())
```

```python
df_clean["date_added"].str.match(r"^\d{1,2}/\d{1,2}/\d{4}$").value_counts()
```

```python
parts = df_clean["date_added"].str.split("/", expand=True).astype(int)
print(parts[0].max(), parts[1].max())
```

```python
df_clean["duration"].str.split(" ").str[1].value_counts()
pd.crosstab(df_clean["type"], df_clean["duration"].str.split(" ").str[1])
```

```python
df_clean["date_added"] = pd.to_datetime(df_clean["date_added"], format="%m/%d/%Y")
```

```python
df_clean["duration_value"] = df_clean["duration"].str.split(" ").str[0].astype(int)
df_clean["duration_unit"] = df_clean["duration"].str.split(" ").str[1].replace("Seasons", "Season")
```

```python
print(df_clean[["date_added", "duration_value"]].dtypes)
print(df_clean["date_added"].min(), df_clean["date_added"].max())
print(df_clean.groupby("duration_unit")["duration_value"].agg(["count", "min", "max"]))
print(df_clean.shape)
```

```python
year_added = df_clean["date_added"].dt.year
early = df_clean[year_added < df_clean["release_year"]]
print(len(early))
print((year_added - df_clean["release_year"]).describe())
early[["title", "type", "release_year", "date_added"]].head(10)
```
