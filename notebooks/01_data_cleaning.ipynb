### Decision: possibly auto-converted titles
Found 5 titles matching a date-like pattern: 9-Feb, 15-Aug, 22-Jul, SPF-18, 1-Oct.
SPF-18 looks like a genuine title (false positive). The other 4 were probably
converted from names such as "22 July" by spreadsheet software. The original
names cannot be recovered from the file alone, so titles were left unchanged and
the issue is documented as a limitation. Impact: 4 of 8,787 rows (about 0.05%).

### Decision: near-duplicate titles
Three titles (Love in a Puff, Esperando la carroza, Sin senos sí hay paraíso)
appear twice with the same title, director, country, year, duration and genres,
but different date_added (and a different rating for one). They are not exact
duplicates and the cause cannot be determined from the file, so all six rows
were kept. Maximum effect: 3 of 8,786 titles (about 0.03%).
One exact duplicate (Consequences, s6530) was removed after trimming a stray
space, which made it identical to s3372.

```python
df_clean["main_country"] = df_clean["country"].str.split(",").str[0].str.strip()
df_clean["n_countries"] = df_clean["country"].str.split(",").str.len()
df_clean["main_genre"] = df_clean["listed_in"].str.split(",").str[0].str.strip()
df_clean["n_genres"] = df_clean["listed_in"].str.split(",").str.len()
df_clean["year_added"] = df_clean["date_added"].dt.year
df_clean["month_added"] = df_clean["date_added"].dt.month_name()
```
